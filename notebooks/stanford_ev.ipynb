# Stanford passenger EV BMS field data

This notebook lists what the release holds: the vehicle, its sessions, the signals and the laboratory files. The release is licensed CC BY-NC-ND 4.0, so the notebook shows no values or plots drawn from the data. It makes no diagnosis.

## 1. Released unit

This step lists what `systems('stanford_ev')` reports for the release.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import pandas as pd

ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'fielddata').exists())
sys.path.insert(0, str(ROOT))
from fielddata import load, systems

released = systems('stanford_ev')
display(released)
```

The release is one passenger EV, with its charging and driving sessions and their first and last start time. The record does not name the vehicle.

## 2. Field sessions

This step lists the field session files, one per charging or driving session, named by their start time.

```python
from fielddata.loaders import stanford_ev
sessions = stanford_ev.sessions()
print(sessions['kind'].value_counts().to_dict())
display(sessions[['session', 'kind', 'start']].head())
```

There are 4,322 driving and 1,094 charging sessions, from 21 September 2021 to 9 April 2025.

## 3. Signals

This step lists the signals each session file holds, from the loader's signal list.

```python
print(stanford_ev.SIGNALS)
```

Each session holds pack current, voltage and SOC, 108 cell voltages, 72 module temperatures and pack temperature, each with its own time base. `load(session, signal)` reads one of them from the original file.

## 4. Laboratory files

This step lists the laboratory files by campaign.

```python
lab = stanford_ev.lab_files()
display(lab.groupby('campaign').size())
```

The two laboratory campaigns hold 48 and 52 module files. Five files of the first campaign are empty, as released.
